## 2. Introducción a Numpy

### ¿Qué es Numpy?

**NumPy (Numerical Python)** es la librería fundamental para el cálculo científico en Python.  
Fue creada a principios de los 2000 como evolución de Numeric y numarray, con el objetivo de dotar a Python de:
- Operaciones numéricas eficientes.  
- Estructuras de datos similares a las de **MATLAB** y **Fortran**, pero integradas en Python.  
- Un ecosistema base sobre el que otras librerías pudieran construir (ej. Pandas, Scikit-learn, PyTorch, TensorFlow).  

Hoy en día, **NumPy es la base de casi toda la computación científica y de IA en Python**.

### ¿Por qué es tan utilizado?

1. **Rendimiento**  
   - Sus operaciones están implementadas en **C** y **Fortran**, mucho más rápidas que los bucles en Python puro.  
   - Permite trabajar con **arrays completos de datos** en vez de elementos individuales (vectorización).  

2. **Ecosistema**  
   - Librerías como **Pandas, Scikit-learn, PyTorch, TensorFlow, OpenCV, Matplotlib** se apoyan en NumPy.  
   - Se ha convertido en un **estándar de facto** en el mundo científico y de la IA.  

3. **Flexibilidad**  
   - Admite desde operaciones matemáticas simples hasta álgebra lineal avanzada, transformadas de Fourier o generación de números aleatorios.  

4. **Compatibilidad**  
   - Es el **lenguaje común** para intercambiar datos entre librerías de IA: casi todas aceptan y devuelven objetos de tipo NumPy.

### 📌 ¿Qué tiene por debajo? (Memoria y ndarray)

- Los arrays de NumPy son **bloques contiguos de memoria**, similares a los de C.  
- Esto permite:  
  - Accesos mucho más rápidos que las listas de Python.  
  - Operaciones vectorizadas: aplicar cálculos a millones de elementos en paralelo sin necesidad de bucles.  
- Internamente, un array NumPy se define con:
  - **Un puntero** a la memoria.  
  - **Metadatos** (forma, tamaño, tipo de dato).  
  - **El tipo de datos (`dtype`)** de los elementos (int8, float32, etc.).  

Esta simplicidad de diseño es lo que lo hace tan eficiente.

### 📌 Comparativa con listas nativas de Python

Las listas en Python son estructuras **flexibles y heterogéneas**:  
- Pueden contener elementos de distinto tipo.  
- Cada elemento es un **objeto** con metadatos adicionales.  
- Esto las hace mucho más lentas para operaciones numéricas masivas.  

En cambio, un **array de NumPy**:
- Es homogéneo (todos los elementos del mismo tipo).  
- Está almacenado en memoria contigua.  
- Aprovecha implementaciones en C para operaciones vectorizadas.

Conclusión:
- **Listas**: muy flexibles, ideales para colecciones generales.  
- **NumPy arrays**: especializados en cálculo numérico, mucho más eficientes en memoria y velocidad.  

En IA y ciencia de datos, donde trabajamos con millones de valores, **la diferencia es abismal**.

### ¿Para qué sirve NumPy en IA?

- Representación de **datos y features** en forma de vectores y matrices.  
- Implementación de operaciones de **álgebra lineal**: producto escalar, matrices, autovalores, descomposiciones.  
- Base para los **tensores** en frameworks de Deep Learning (TensorFlow, PyTorch).  
- Procesamiento de señales e imágenes (arrays de píxeles).  
- Generación de datos sintéticos para entrenar modelos.  

En resumen: **sin NumPy, el ecosistema moderno de IA en Python no existiría**.

### 📌 <u>Comparaciones y Máscaras booleanas</u>

Una **máscara booleana** es simplemente un array de `True`/`False` que tiene la misma forma que el array original.  
Se genera aplicando una **comparación** sobre el array.  

Esta máscara puede usarse para **filtrar** elementos, seleccionar subconjuntos o incluso **modificar valores**.

````python
x = np.arange(10)
print("Array original:", x)

# Comparaciones → producen arrays booleanos
print("x > 5 :", x > 5)
print("x % 2 == 0 :", x % 2 == 0)
````

> **Filtrado con máscaras**
>-  Podemos usar la máscara como índice para extraer solo los elementos que cumplen la condición.

````python
# Extraer solo los mayores que 5
mask = x > 5
print("Máscara:", mask)
print("Elementos mayores que 5:", x[mask])

# En una sola línea
print("Elementos pares:", x[x % 2 == 0])
````

> **Combinación de condiciones**
>-  Al igual que en Pandas, se pueden usar operadores lógicos:
>>-  `&` (AND)
>>-  `|` (OR)
>>-  `~` (NOT)
>-  **Importante**: cada condición debe ir entre paréntesis.

````python
# Números entre 3 y 7
print("Entre 3 y 7:", x[(x >= 3) & (x <= 7)])

# Números menores que 2 o mayores que 8
print("Fuera de [2,8]:", x[(x < 2) | (x > 8)])

# Negación: NO pares
print("Impares:", x[~(x % 2 == 0)])
````

### 📝 **Ejercicios**

1. Crea un array 2D de forma `(5, 5)` con números aleatorios entre 0 y 50.
2. Normalízalo al rango [0, 1] restando el mínimo y dividiendo entre el máximo ((valor - min) / (max - min)).  
3. Usa una **máscara booleana** para extraer únicamente los elementos mayores a 0.5.  
4. (Extra) **Binariza** el array normalizado:  
   - Los valores > 0.5 deben pasar a 1.  
   - Los valores ≤ 0.5 deben pasar a 0.  
   - Hazlo de dos formas:  
     - Usando una máscara booleana.  
     - Usando `np.where`.

````python
import numpy as np

arr = np.random.randint(0, 51, size=(5, 5))
print("Array original:\n", arr)

arr_min = arr.min()
arr_max = arr.max()
arr_norm = (arr - arr_min) / (arr_max - arr_min)
print("\nArray normalizado:\n", arr_norm)

mask = arr_norm > 0.5
print("\nMáscara booleana:\n", mask)

valores_mayores = arr_norm[mask]
print("\nElementos mayores a 0.5:\n", valores_mayores)

# --- Forma 1: usando una máscara booleana
arr_bin_mask = arr_norm.copy()
arr_bin_mask[arr_norm > 0.5] = 1
arr_bin_mask[arr_norm <= 0.5] = 0
print("\nBinarización (con máscara):\n", arr_bin_mask)

# --- Forma 2: usando np.where
arr_bin_where = np.where(arr_norm > 0.5, 1, 0)
print("\nBinarización (con np.where):\n", arr_bin_where)
````

Los arrays de NumPy permiten acceder a subconjuntos de datos de forma **muy eficiente**.

### 📌 <u>Indexación simple y por rangos (`:`)</u>

- Igual que en listas de Python: `arr[i]` devuelve el elemento en posición *i*.  
- Con el operador `:` podemos seleccionar rangos de posiciones (`start:stop:step`).  
- En arrays multidimensionales, se usa una coma para separar cada dimensión.

````python
arr = np.arange(10)
print("Array:\n", arr)
print("\n")
print("Elemento en posición 3:", arr[3])
print("Rango del 2 al 6:", arr[2:7])
print("Cada dos elementos:", arr[::2])

print("\n")
print("-"*100)
print("\n")

# En 2D
M = np.arange(12).reshape(3, 4)
print("Matriz:\n", M)
print("\n")
print("Elemento [1,2]:", M[1, 2])
print("Fila 1:", M[1, :])
print("Columna 2:", M[:, 2])
````

### <u>Ejemplos con extracción de submatrices</u>

Podemos usar slicing para obtener sub-bloques de una matriz.

````python
M = np.arange(25).reshape(5, 5)
print("Matriz 5x5:\n", M)
print("\n")
print("Submatriz central (3x3):\n", M[1:4, 1:4])
print("\n")
print("Fila 2, columnas 1 a 3:\n", M[2, 1:4])
print("\n")
print("Esquinas:\n", M[[0, -1], :][:, [0, -1]])
````

### 📌 <u>Diferencia entre vista y copia</u>

- **Cuando haces slicing (`arr[1:5]`), NumPy devuelve una vista del array original, no una copia.**
- Esto significa que si modificas la vista, cambian también los valores del array original.
- Esto es muy distinto de lo que pasa en listas de Python.

````python
x = np.arange(10)
print(x)

sub = x[2:5]
print(sub)

sub[:] = -1
print("Original:", x)   # también se modifica

print("\n")

## SI QUIERES FORZAR UNA COPIA
y = np.arange(10)
print(y)

sub_copy = y[2:5].copy()
print(sub_copy)

sub_copy[:] = -1
print("Original:", y)   # no cambia
print("sub_copy:", sub_copy)
````

### <u>Función `np.ix_` para combinaciones</u>

## 5. Broadcasting

El **broadcasting** permite realizar operaciones entre arrays de **formas distintas**, siempre que cumplan ciertas reglas de compatibilidad.

En lugar de copiar datos o escribir bucles, NumPy **"expande" virtualmente** los arrays más pequeños para que coincidan en dimensiones con los más grandes.

Internamente:  
- NumPy compara las **formas** de los arrays.  
- Donde hay un `1`, interpreta que esa dimensión puede repetirse tantas veces como sea necesario.  
- No se crean copias físicas de los datos: la expansión es **virtual** (muy eficiente en memoria).  

Esto hace el código:
- Más **rápido** (operaciones vectorizadas en C).  
- Más **legible** (sin bucles explícitos).  
- Más **eficiente en memoria** (no duplica datos, solo simula la expansión).

### 📌 <u>Reglas de Broadcasting</u>

Cuando NumPy encuentra dos arrays con formas diferentes:

1. Compara las dimensiones de **derecha a izquierda**.  
2. Dos dimensiones son **compatibles** si:  
   - Son iguales, o  
   - Una de ellas es 1.  
3. Si todas las dimensiones son compatibles, se expanden virtualmente.  
4. Si alguna no lo es, se lanza un error.

<u>Nota:</u> Cuando una dimensión es 1, NumPy **estira virtualmente** el array en esa dimensión para que encaje.

### <u>Ejemplo simple: matriz + vector</u>

````python
import numpy as np

M = np.arange(12).reshape(3, 4)
v = np.array([10, 20, 30, 40])

print("Matriz:\n", M)
print("\n")
print("Vector:\n", v)

print("\n")
resultado = M + v
print("Resultado:\n", resultado)
````

**¿Qué hace NumPy por debajo?**

- Forma de `M`: (3, 4)  
- Forma de `v`: (4,) → NumPy lo interpreta como (1, 4).  
- Comparación:  
  - Dimensión 1: 4 vs 4 → ✅ compatibles.  
  - Dimensión 0: 3 vs 1 → ✅ se estira el `1` a `3`.  
- Forma resultante: (3, 4).  

El vector `v` se **replica virtualmente** en cada fila de la matriz.

### 📌 <u>Ejemplo no trivial: compatibilidad de dimensiones</u>

````python
A = np.ones((3, 1))
print("Vector A:\n", A)
print("\n")

B = np.arange(4)
print("Vector B:\n", B)

print("\n")
C = A + B
print("Resultado (forma):", C.shape)
print(C)
````

**¿Qué hace NumPy por debajo?**

- Forma de `A`: (3, 1)  
- Forma de `B`: (4,) → se interpreta como (1, 4).  
- Comparación:  
  - Dimensión 1: 1 vs 4 → ✅ el `1` se expande a `4`.  
  - Dimensión 0: 3 vs 1 → ✅ el `1` se expande a `3`.  
- Forma resultante: (3, 4).  

**`A` se repite en columnas, `B` se repite en filas.**   
El resultado es una matriz 3x4.

### <u>Aplicación práctica: normalización de columnas</u>

````python
X = np.random.randint(0, 10, (5, 3))  # forma (5,3)
col_means = X.mean(axis=0)            # forma (3,)

print("Matriz original:\n", X)
print("Medias por columna:", col_means)

X_norm = X - col_means
print("Matriz normalizada:\n", X_norm)
````

**¿Qué hace NumPy por debajo?**

- Forma de `X`: (5, 3).  
- Forma de `col_means`: (3,) → se interpreta como (1, 3).  
- Comparación:  
  - Dimensión 1: 3 vs 3 → ✅ compatibles.  
  - Dimensión 0: 5 vs 1 → ✅ se estira el `1` a `5`.  
- Forma resultante: (5, 3).  

**Cada columna de `X` resta su propia media sin necesidad de bucles.**

### <u>Notas avanzadas de Broadcasting</u>

Ya hemos visto cómo funciona el broadcasting en casos simples y prácticos.  
Ahora veremos algunos matices y ejemplos avanzados.

- **1. Eficiencia de memoria:**

Cuando NumPy expande una dimensión de tamaño `1`, **no copia los datos**, solo crea una vista virtual.  
Esto evita desperdiciar memoria y hace que el broadcasting sea muy eficiente.

- **2. Broadcasting en comparaciones:**

El broadcasting también funciona con comparaciones, lo que permite crear **máscaras booleanas** de forma muy flexible.

````python
X = np.arange(12).reshape(3, 4)
thresholds = np.array([2, 2, 5, 5])  # (4,)

mask = X > thresholds
print("Matriz:\n", X)
print("\n")
print("thresholds:\n", thresholds)
print("\n")
print("Máscara:\n", mask)
````

- **3. Broadcasting con más de dos arrays:**

Las reglas de broadcasting se aplican cuando hay **más de dos arrays**.  
NumPy busca una forma común compatible para todos ellos.

````python
A = np.ones((3, 1, 4))
B = np.ones((1, 5, 1))

C = A + B
print("Formas: A", A.shape, "B", B.shape, "→ C", C.shape) 
````

- **4. Cuando el broadcasting falla:**

Si las dimensiones no son compatibles y ninguna es `1`, NumPy lanza un error.

````python
A = np.ones((3, 4))
B = np.ones((2, 4))

A + B
````

## 6. Algebra lineal en NumPy

Muchas operaciones fundamentales en Inteligencia Artificial (entrenamiento de redes, regresión lineal, optimización, etc.) se basan en **álgebra lineal**.  

### 📌 <u>Producto punto y multiplicación matricial (`np.dot` o `@`)</u>

El producto punto es la base de muchas operaciones:  
- Entre vectores: combinación lineal.  
- Entre matrices: suma de productos fila-columna.  

En Python, además de `np.dot(a, b)`, podemos usar el operador `@`.

````python
## VECTORES

a = np.array([1, 2, 3])
print("a:", a)
b = np.array([4, 5, 6])
print("b:", b)

dot1 = np.dot(a, b)
dot2 = a @ b

print("np.dot:", dot1)
print("a @ b :", dot2)
````

````python
## MATRICES

A = np.random.randint(0, 5, (2, 3))
B = np.random.randint(0, 5, (3, 2))
print("A:\n", A)
print("B:\n", B)

dot1 = np.dot(A, B)
dot2 = A @ B

print("np.dot:\n", dot1)
print("A @ B :\n", dot2)
````

### <u>Transposición (`.T`)</u>

La transposición cambia filas por columnas, algo esencial en productos matriciales.

````python
M = np.array([[1, 2, 3],
              [4, 5, 6]])

print("Matriz:\n", M)
print("Transpuesta:\n", M.T)
````

### <u>Norma (`np.linalg.norm`)</u>

La norma mide la **longitud** de un vector desde el origen. Por defecto calcula la norma Euclídea (L2).  
La norma L2 se calcula como la suma de las componentes del vector al cuadrado y, posteriormente, aplicar la raíz cuadrada al resultado

````python
v = np.array([3, 4])
print("Vector:", v)
print("Norma L2:", np.linalg.norm(v))
````

### <u>Determinante de matrices (`np.linalg.det(A)`)</u>

El determinante de una matriz es **un número (o escalar)** que se obtiene para matrices cuadradas  
y proporciona información importante sobre la matriz, como si el sistema de ecuaciones lineales que representa tiene solución única.

- Propiedades de los determinantes:
>- Un sistema de ecuaciones lineales tiene una solución única si el determinante de la matriz de coeficientes es distinto de cero.
>- Una matriz tiene inversa si y solo si su determinante es diferente de cero. 

````python
A = np.array([[4, 7],
              [2, 6]])

print(f"A:\n{A}")

A_det = np.linalg.det(A)
print("Determinante de A:\n", A_det)
````

### <u>Inversa de matrices (`np.linalg.inv`)</u>

La matriz inversa es aquella que cumple:  
\[
A · A⁻¹ = I
\]

<u>NOTA:</u> Solo existe si la matriz es **cuadrada y no singular** (determinante ≠ 0).

````python
A = np.array([[4, 7],
              [2, 6]])

A_inv = np.linalg.inv(A)
print("Inversa de A:\n", A_inv)

# Comprobación: A @ A_inv ≈ I
print("A @ A_inv ≈\n", A @ A_inv)
````

### <u>Resolución de sistemas lineales (`np.linalg.solve`)</u>

Podemos resolver sistemas de ecuaciones lineales de la forma:

\[
A · x = b
\]

donde:  
- \(A\) es la matriz de coeficientes.  
- \(x\) es el vector de incógnitas.  
- \(b\) es el vector de resultados.

````python
# Sistema:
# 2x + y - z =  8
# -3x - y + 2z = -11
# -2x + y + 2z = -3

A = np.array([[ 2,  1, -1],
              [-3, -1,  2],
              [-2,  1,  2]])

b = np.array([ 8, -11, -3])

x = np.linalg.solve(A, b)
print("Solución del sistema:", x)
````

### <u>Resumen de Álgebra Lineal en NumPy</u>

| Operación                        | Descripción                               | Función NumPy / Sintaxis       |
|----------------------------------|-------------------------------------------|--------------------------------|
| Producto punto (vectores)        | Suma de productos de los elementos         | `np.dot(a, b)` ó `a @ b`       |
| Multiplicación de matrices       | Producto fila-columna entre matrices       | `A @ B` ó `np.matmul(A, B)`    |
| Transpuesta                      | Intercambio de filas y columnas            | `A.T`                          |
| Norma (longitud vector)          | Magnitud de un vector (por defecto L2)     | `np.linalg.norm(a)`            |
| Determinante                     | Valor que caracteriza a una matriz         | `np.linalg.det(A)`             |
| Inversa de una matriz            | Matriz que cumple A · A⁻¹ = I              | `np.linalg.inv(A)`             |
| Sistema de ecuaciones lineales   | Resolver A · x = b                         | `np.linalg.solve(A, b)`        |
| Autovalores y autovectores       | Descomposición A v = λ v                   | `np.linalg.eig(A)`             |
| Descomposición en valores sing.  | Factorización A = U Σ Vᵀ                   | `np.linalg.svd(A)`             |

## 7. `.to_numpy()` y `.values`

- `.to_numpy()` → la forma **recomendada** y moderna de convertir un DataFrame o Series a `ndarray`.  
- `.values` → hace lo mismo, pero es menos explícito y puede dar sorpresas con tipos más complejos (ej. categóricos).  

Ambos devuelven un `numpy.ndarray`.

````python
import pandas as pd
import numpy as np

# Creamos un DataFrame
df = pd.DataFrame({
    "A": [1, 2, 3, 4],
    "B": [10, 20, 30, 40],
    "C": [5.5, 6.5, 7.5, 8.5]
})

print("DataFrame:\n", df)

# Convertir a NumPy
arr1 = df.to_numpy()
arr2 = df.values

print("\n.to_numpy():\n", arr1, type(arr1))
print("\n.values:\n", arr2, type(arr2))
````

### <u>Uso combinado para cálculos vectorizados</u>

Imagina que tenemos datos en un DataFrame, pero queremos hacer cálculos matemáticos avanzados (ej. normalización, distancias, transformaciones) usando NumPy.

````python
# Seleccionamos solo las columnas numéricas
X = df[["A", "B", "C"]].to_numpy()

# Normalizar cada columna (broadcasting en NumPy)
col_means = X.mean(axis=0)
col_stds = X.std(axis=0)

X_norm = (X - col_means) / col_stds

print("Array normalizado (NumPy):\n", X_norm)
````

### <u>Ventajas de la integración</u>

- **De Pandas a NumPy**: fácil acceso a cálculos matemáticos avanzados.  
- **De NumPy a Pandas**: podemos reconstruir DataFrames con etiquetas de columnas.  

Esto nos da lo mejor de ambos mundos:
- Pandas → manipulación y limpieza de datos con etiquetas.  
- NumPy → operaciones matemáticas rápidas y vectorizadas.

````python
# Volvemos a DataFrame, manteniendo nombres de columnas
df_norm = pd.DataFrame(X_norm, columns=["A", "B", "C"])
print(df_norm)
````

***20. Compara los tiempos de ejecución y explica la diferencia.***

La enorme diferencia de tiempo en la ejecución (que hemos comprobado con `%timeit`) se debe a que la versión vectorizada delega las operaciones a rutinas en **lenguaje C altamente optimizadas**, mientras que el bucle procesa los datos uno a uno a través del intérprete de Python.

Las tres razones técnicas principales son:

1. **El cuello de botella del intérprete (Overhead de Python):**
   En el bucle `for`, Python evalúa el código línea por línea. En cada iteración tiene que comprobar qué tipo de dato es el número (tipado dinámico), buscarlo en la memoria, calcular y asignar el resultado. Este proceso añade un retraso enorme cuando se repite miles de veces.

2. **Gestión de memoria contigua:**
   En la versión vectorizada, NumPy sabe que toda tu matriz contiene un único tipo de dato (ej. `float64`) almacenado en un bloque de memoria continuo. Envía todo ese bloque directamente a sus motores internos en C, eliminando por completo las comprobaciones repetitivas de tipo y las búsquedas fragmentadas en memoria.

3. **Paralelización de hardware (Instrucciones SIMD):**
   NumPy aprovecha las instrucciones SIMD (*Single Instruction, Multiple Data*) de los procesadores modernos. Esto significa que la CPU de tu ordenador puede multiplicar varios pares de números de tu matriz exactamente en el mismo ciclo de reloj, en lugar de hacerlo de uno en uno como te obliga el bucle.

> **Conclusión:** La regla de oro en el análisis de datos es evitar los bucles tradicionales siempre que sea posible. Delegar el cálculo a operaciones vectorizadas hace el código más limpio y, sobre todo, órdenes de magnitud más rápido.
